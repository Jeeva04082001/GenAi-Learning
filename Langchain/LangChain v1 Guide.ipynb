# Create Agent

```python
from langchain.agents import create_agent

agent = create_agent(model="groq:openai/gpt-oss-120b")

response = agent.invoke({"messages":"what is yoor name"})

print(response)



```

```python
from pprint import pprint
pprint(response)
```

```python
response["messages"][-1].content
```

# Stream Method

```python
stream_response = agent.stream({"messages":"what is your name,tell me 20 jokes"},stream_mode="messages")

```

```python
for token,j in stream_response:
    print(token.content,end="",flush=True)
```

# Batch Method

```python
bath_response = agent.batch([
    {"messages":"what is your name"},
    {"messages":"who created you"}
])

```

```python
bath_response
```

```python
for i in bath_response:
    print(i["messages"][-1].content)
    print("------------------------")
    
```

# Messages

- system message - prompt message (system prompt)
- human message - user questions
- tool message - tool response
- ai message - llm answer

# Prompt


```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
template = ChatPromptTemplate(
    [
        ("system", "You are a helpful AI bot. Your name is Sam."),
        ("human", "Hello, how are you doing?"),
        ("ai", "I'm doing well, thanks!"),
        ("human", "{user_input}"),
    ]
)

prompt_value = template.invoke(
    {
        "user_input": "What is your name?",
    }
)

```

```python
pprint(prompt_value)
```

```python
agent.invoke(prompt_value)
```

```python
agent.invoke({"messages":"what is your name"})
```

```python
system_prompt=""" 

Your name is sam, your an chatbot assistant 
woking in ABC-fintech, company phone numbers is +987877799
email is "abc@gmail.com"

"""
```

```python
system_prompt_agent = create_agent(
    model="groq:openai/gpt-oss-120b",system_prompt=system_prompt
)
```

```python
response = system_prompt_agent.invoke({"messages":"who are you, What is your company name"})
```
